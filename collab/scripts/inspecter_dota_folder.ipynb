**Dictionnaire de mapping et nouvelle taxonomie**

```python
dota_names = ['small-vehicle', 'large-vehicle', 'plane', 'storage-tank', 'ship',
              'harbor', 'ground-track-field', 'soccer-ball-field', 'tennis-court',
              'swimming-pool', 'baseball-diamond', 'roundabout', 'basketball-court',
              'bridge', 'helicopter']

xview_names = data_yaml_merged["names"]  # les 29 classes déjà chargées

nouvelles_classes = ['Harbor', 'Ground Track Field', 'Soccer Ball Field', 'Tennis Court',
                      'Swimming Pool', 'Baseball Diamond', 'Roundabout',
                      'Basketball Court', 'Bridge']

noms_finaux = xview_names + nouvelles_classes
print(f"Taxonomie finale : {len(noms_finaux)} classes")

dota_to_final = {
    0: xview_names.index('Small Car'),
    1: xview_names.index('Truck'),
    2: xview_names.index('Aircraft'),
    3: xview_names.index('Storage Tank'),
    4: xview_names.index('Boat'),
    5: noms_finaux.index('Harbor'),
    6: noms_finaux.index('Ground Track Field'),
    7: noms_finaux.index('Soccer Ball Field'),
    8: noms_finaux.index('Tennis Court'),
    9: noms_finaux.index('Swimming Pool'),
    10: noms_finaux.index('Baseball Diamond'),
    11: noms_finaux.index('Roundabout'),
    12: noms_finaux.index('Basketball Court'),
    13: noms_finaux.index('Bridge'),
    14: xview_names.index('Helicopter'),
}
print(dota_to_final)
```

```python
def remapper_labels(dossier_labels_src, dossier_labels_dst, mapping):
    os.makedirs(dossier_labels_dst, exist_ok=True)
    n_fichiers, n_lignes = 0, 0
    for fichier in os.listdir(dossier_labels_src):
        if not fichier.endswith('.txt'):
            continue
        chemin_src = os.path.join(dossier_labels_src, fichier)
        chemin_dst = os.path.join(dossier_labels_dst, fichier)
        with open(chemin_src) as f_in, open(chemin_dst, 'w') as f_out:
            for ligne in f_in:
                parties = ligne.strip().split()
                if not parties:
                    continue
                ancien_id = int(parties[0])
                nouvel_id = mapping[ancien_id]
                f_out.write(f"{nouvel_id} {' '.join(parties[1:])}\n")
                n_lignes += 1
        n_fichiers += 1
    return n_fichiers, n_lignes

for split in ['train', 'val']:
    src = os.path.join(dota_root, split, 'labels')
    dst = os.path.join(dota_root, split, 'labels_remapped')
    n_f, n_l = remapper_labels(src, dst, dota_to_final)
    print(f"{split} : {n_f} fichiers remappés, {n_l} objets réindexés")
```
